```python
# ==============================
# ANÁLISIS DE FILAS ANÓMALAS
# ==============================

# 1) Máscaras base
A = dfc['UnitPrice'] == 0
B = dfc['Quantity'] <= 0
C = dfc['CustomerID'].isnull()
D = dfc['Description'].isnull()

# 2) Tres opciones de filtrado
mask1 = (A & B) | (C & D)
mask2 = A & B & C & D
mask3 = A | (B & C & D)

# 3) Mostrar cuántas filas devuelve cada opción
print("RESULTADOS DE LAS 3 OPCIONES")
print(f"Opción 1 -> ((UnitPrice == 0) y (Quantity <= 0)) o ((CustomerID nulo) y (Description nula)): {mask1.sum()}")
print(f"Opción 2 -> (UnitPrice == 0) y (Quantity <= 0) y (CustomerID nulo) y (Description nula): {mask2.sum()}")
print(f"Opción 3 -> (UnitPrice == 0) o ((Quantity <= 0) y (CustomerID nulo) y (Description nula)): {mask3.sum()}")

# 4) Elegir qué opción analizar
# Cambia aquí: "opcion1", "opcion2" o "opcion3"
modo_analisis = "opcion3"

if modo_analisis == "opcion1":
    condiciones_especificas = mask1
    descripcion_modo = "((UnitPrice == 0) y (Quantity <= 0)) o ((CustomerID nulo) y (Description nula))"
elif modo_analisis == "opcion2":
    condiciones_especificas = mask2
    descripcion_modo = "(UnitPrice == 0) y (Quantity <= 0) y (CustomerID nulo) y (Description nula)"
elif modo_analisis == "opcion3":
    condiciones_especificas = mask3
    descripcion_modo = "(UnitPrice == 0) o ((Quantity <= 0) y (CustomerID nulo) y (Description nula))"
else:
    raise ValueError("modo_analisis debe ser: 'opcion1', 'opcion2' o 'opcion3'")

# 5) Subconjunto filtrado
filas_caso_especifico = dfc[condiciones_especificas].copy()

print("\n" + "="*70)
print(f"MODO DE ANÁLISIS SELECCIONADO: {modo_analisis}")
print(f"Lógica aplicada: {descripcion_modo}")
print(f"Número de registros encontrados: {len(filas_caso_especifico)}")

if filas_caso_especifico.empty:
    print("No se encontraron filas que cumplan con las condiciones específicas.")
else:
    # 6) Crear banderas de anomalía dentro del subconjunto filtrado
    filas_caso_especifico['anomalia_unitprice_0'] = filas_caso_especifico['UnitPrice'] == 0
    filas_caso_especifico['anomalia_quantity_le_0'] = filas_caso_especifico['Quantity'] <= 0
    filas_caso_especifico['anomalia_quantity_lt_0'] = filas_caso_especifico['Quantity'] < 0
    filas_caso_especifico['anomalia_quantity_eq_0'] = filas_caso_especifico['Quantity'] == 0
    filas_caso_especifico['anomalia_customerid_nulo'] = filas_caso_especifico['CustomerID'].isnull()
    filas_caso_especifico['anomalia_description_nula'] = filas_caso_especifico['Description'].isnull()

    # 7) Etiqueta exacta de combinación de anomalías
    def etiquetar_fila(row):
        etiquetas = []
        if row['anomalia_unitprice_0']:
            etiquetas.append("UnitPrice=0")
        if row['anomalia_quantity_le_0']:
            etiquetas.append("Quantity<=0")
        if row['anomalia_customerid_nulo']:
            etiquetas.append("CustomerID nulo")
        if row['anomalia_description_nula']:
            etiquetas.append("Description nula")
        return " + ".join(etiquetas) if etiquetas else "Sin anomalía"

    filas_caso_especifico['tipo_anomalia'] = filas_caso_especifico.apply(etiquetar_fila, axis=1)

    # 8) Resumen individual por anomalía
    resumen_individual = pd.DataFrame({
        'Métrica': [
            'Total filas filtradas',
            'UnitPrice = 0',
            'Quantity <= 0',
            'Quantity < 0 (devoluciones o ajustes)',
            'Quantity = 0',
            'CustomerID nulo',
            'Description nula',
            'UnitPrice = 0 y Quantity <= 0',
            'CustomerID nulo y Description nula',
            'Las 4 condiciones a la vez'
        ],
        'Cantidad': [
            len(filas_caso_especifico),
            filas_caso_especifico['anomalia_unitprice_0'].sum(),
            filas_caso_especifico['anomalia_quantity_le_0'].sum(),
            filas_caso_especifico['anomalia_quantity_lt_0'].sum(),
            filas_caso_especifico['anomalia_quantity_eq_0'].sum(),
            filas_caso_especifico['anomalia_customerid_nulo'].sum(),
            filas_caso_especifico['anomalia_description_nula'].sum(),
            (filas_caso_especifico['anomalia_unitprice_0'] & filas_caso_especifico['anomalia_quantity_le_0']).sum(),
            (filas_caso_especifico['anomalia_customerid_nulo'] & filas_caso_especifico['anomalia_description_nula']).sum(),
            (
                filas_caso_especifico['anomalia_unitprice_0'] &
                filas_caso_especifico['anomalia_quantity_le_0'] &
                filas_caso_especifico['anomalia_customerid_nulo'] &
                filas_caso_especifico['anomalia_description_nula']
            ).sum()
        ]
    })

    # 9) Resumen por combinaciones exactas
    resumen_combinaciones = (
        filas_caso_especifico['tipo_anomalia']
        .value_counts(dropna=False)
        .reset_index()
    )
    resumen_combinaciones.columns = ['Combinación exacta de anomalías', 'Cantidad']

    # 10) Mostrar resultados
    print("\nRESUMEN INDIVIDUAL DE ANOMALÍAS")
    display(resumen_individual)

    print("\nRESUMEN POR COMBINACIONES EXACTAS")
    display(resumen_combinaciones)

    print("\nPRIMERAS 20 FILAS DETECTADAS")
    display(
        filas_caso_especifico[
            [
                'InvoiceNo', 'StockCode', 'Description', 'Quantity',
                'InvoiceDate', 'UnitPrice', 'CustomerID', 'Country',
                'tipo_anomalia'
            ]
        ].head(20)
    )
```

```python
# ============================================================
# DETECTAR Y ELIMINAR PARES DE CORRECCIÓN / REVERSIÓN
# ------------------------------------------------------------
# Criterio:
# - una fila positiva
# - otra negativa posterior
# - mismo StockCode, Description, UnitPrice, CustomerID, Country
# - misma cantidad en valor absoluto
# - dentro de una ventana temporal configurable
# ============================================================

import pandas as pd
import numpy as np

# -------- PARÁMETRO AJUSTABLE ----------
VENTANA_HORAS = 48   # cambia a 24, 12, 72... según quieras ser más estricto o más flexible
# --------------------------------------

tmp = dfc.copy()

# Guardar índice original para poder borrar después
tmp['idx_original'] = tmp.index

# Asegurar formato datetime
tmp['InvoiceDate'] = pd.to_datetime(tmp['InvoiceDate'], errors='coerce')

# Normalización de claves para emparejar mejor
tmp['StockCode_norm'] = tmp['StockCode'].astype(str).str.strip().str.upper()
tmp['Description_norm'] = tmp['Description'].fillna('').astype(str).str.strip().str.upper()
tmp['Country_norm'] = tmp['Country'].fillna('').astype(str).str.strip().str.upper()

# CustomerID: permitir emparejar también nulos
tmp['CustomerID_norm'] = tmp['CustomerID'].astype(str).replace('nan', '__NULL__')

# Redondeo por seguridad en precios
tmp['UnitPrice_norm'] = tmp['UnitPrice'].round(2)

# Cantidad absoluta
tmp['abs_qty'] = tmp['Quantity'].abs()

# Separar positivos y negativos
pos = tmp[tmp['Quantity'] > 0].copy()
neg = tmp[tmp['Quantity'] < 0].copy()

# Renombrar para merge
pos = pos.rename(columns={
    'idx_original': 'idx_pos',
    'InvoiceDate': 'InvoiceDate_pos',
    'Quantity': 'Quantity_pos',
    'InvoiceNo': 'InvoiceNo_pos'
})

neg = neg.rename(columns={
    'idx_original': 'idx_neg',
    'InvoiceDate': 'InvoiceDate_neg',
    'Quantity': 'Quantity_neg',
    'InvoiceNo': 'InvoiceNo_neg'
})

# Claves de emparejamiento
keys = [
    'StockCode_norm',
    'Description_norm',
    'UnitPrice_norm',
    'CustomerID_norm',
    'Country_norm',
    'abs_qty'
]

# Cruce candidatos
candidatos = pos.merge(
    neg,
    on=keys,
    how='inner',
    suffixes=('_pos', '_neg')
)

# Quedarnos solo con negativos posteriores y dentro de ventana
candidatos = candidatos[
    (candidatos['InvoiceDate_neg'] > candidatos['InvoiceDate_pos']) &
    ((candidatos['InvoiceDate_neg'] - candidatos['InvoiceDate_pos']) <= pd.Timedelta(hours=VENTANA_HORAS))
].copy()

# Diferencia temporal
candidatos['delta_tiempo'] = candidatos['InvoiceDate_neg'] - candidatos['InvoiceDate_pos']

# Ordenar por cercanía temporal
candidatos = candidatos.sort_values(['idx_pos', 'delta_tiempo', 'idx_neg'])

# Emparejamiento 1 a 1:
# 1) primer negativo más cercano para cada positivo
pares = candidatos.drop_duplicates(subset='idx_pos', keep='first').copy()

# 2) asegurar que un mismo negativo no se use dos veces
pares = pares.sort_values(['idx_neg', 'delta_tiempo']).drop_duplicates(subset='idx_neg', keep='first').copy()

# Orden final
pares = pares.sort_values(['InvoiceDate_pos', 'InvoiceDate_neg']).copy()

# Resumen de pares detectados
columnas_resumen = [
    'idx_pos', 'InvoiceNo_pos', 'idx_neg', 'InvoiceNo_neg',
    'StockCode_norm', 'Description_norm',
    'Quantity_pos', 'Quantity_neg',
    'UnitPrice_norm', 'CustomerID_norm', 'Country_norm',
    'InvoiceDate_pos', 'InvoiceDate_neg', 'delta_tiempo'
]

resumen_pares = pares[columnas_resumen].copy()

print("==============================================")
print("ANÁLISIS DE PARES DE CORRECCIÓN / REVERSIÓN")
print("==============================================")
print(f"Ventana temporal usada: {VENTANA_HORAS} horas")
print(f"Pares detectados: {len(resumen_pares)}")
print(f"Filas que se eliminarán: {len(resumen_pares) * 2}")

display(resumen_pares.head(30))

# Índices a eliminar
indices_borrar = pd.Index(pares['idx_pos']).append(pd.Index(pares['idx_neg'])).unique()

# Crear dataset limpio
dfc = dfc.drop(index=indices_borrar, errors='ignore').copy()

print("\n==============================================")
print("LIMPIEZA APLICADA")
print("==============================================")
print(f"Índices eliminados: {len(indices_borrar)}")
print(f"Filas restantes en dfc: {len(dfc)}")

#comaprar df con dfc y ver cuantas lineas se han eliminado
print(f"Filas borradas tras aplicar la regla 4: {len(df) - len(dfc)}")
```

```python
# ============================================================
# EDA FINAL COMPARATIVO: df (original) vs dfc (limpio)
# + opcional: dfc_clustering
# Requiere que ya existan df y dfc en memoria.
# ============================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from IPython.display import display
from matplotlib.ticker import FuncFormatter

# ------------------------------------------------------------
# 0) CONFIGURACIÓN VISUAL
# ------------------------------------------------------------
sns.set_theme(style="whitegrid", context="talk", palette="deep")
plt.rcParams["figure.figsize"] = (14, 7)
plt.rcParams["axes.titlesize"] = 18
plt.rcParams["axes.labelsize"] = 13
plt.rcParams["xtick.labelsize"] = 11
plt.rcParams["ytick.labelsize"] = 11

OUTPUT_DIR = "graficos_eda_final"
os.makedirs(OUTPUT_DIR, exist_ok=True)

def euro_fmt(x, pos):
    return f"{x:,.0f}".replace(",", ".")

def save_show(fig, filename=None):
    if filename:
        fig.savefig(os.path.join(OUTPUT_DIR, filename), dpi=180, bbox_inches="tight")
    plt.show()

# ------------------------------------------------------------
# 1) COPIAS DE TRABAJO Y NORMALIZACIÓN
# ------------------------------------------------------------
raw = df.copy()
clean = dfc.copy()

datasets = {
    "Original (df)": raw,
    "Limpio (dfc)": clean
}

if "dfc_clustering" in globals():
    datasets["Clustering (dfc_clustering)"] = dfc_clustering.copy()

for name, data in datasets.items():
    # Tipos
    data["InvoiceDate"] = pd.to_datetime(data["InvoiceDate"], errors="coerce")
    data["Quantity"] = pd.to_numeric(data["Quantity"], errors="coerce")
    data["UnitPrice"] = pd.to_numeric(data["UnitPrice"], errors="coerce")
    data["CustomerID"] = pd.to_numeric(data["CustomerID"], errors="coerce")

    # Textos
    for col in ["InvoiceNo", "StockCode", "Description", "Country"]:
        if col in data.columns:
            data[col] = data[col].astype(str).str.strip()

    # Total línea
    data["TotalPrice"] = data["Quantity"] * data["UnitPrice"]

    # Máscaras auxiliares
    data["is_positive_sale"] = (data["Quantity"] > 0) & (data["UnitPrice"] > 0)
    data["is_return"] = (data["Quantity"] < 0) | (data["InvoiceNo"].astype(str).str.upper().str.startswith("C", na=False))
    data["YearMonth"] = data["InvoiceDate"].dt.to_period("M").astype(str)
    data["Date"] = data["InvoiceDate"].dt.date

# ------------------------------------------------------------
# 2) FUNCIONES AUXILIARES
# ------------------------------------------------------------
def iqr_outlier_pct(series):
    s = pd.to_numeric(series, errors="coerce").dropna()
    if len(s) < 5:
        return np.nan
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    return ((s < low) | (s > high)).mean() * 100

def build_summary(name, data):
    sales = data[data["is_positive_sale"]].copy()

    return {
        "Dataset": name,
        "Filas": len(data),
        "Columnas": data.shape[1],
        "Duplicados exactos": int(data.drop(columns=["is_positive_sale", "is_return"], errors="ignore").duplicated().sum()),
        "Nulos Description": int(data["Description"].isna().sum() if data["Description"].dtype != object else (data["Description"].replace("nan", np.nan).isna().sum())),
        "Nulos CustomerID": int(data["CustomerID"].isna().sum()),
        "Quantity <= 0": int((data["Quantity"] <= 0).sum()),
        "UnitPrice <= 0": int((data["UnitPrice"] <= 0).sum()),
        "Filas devoluciones": int(data["is_return"].sum()),
        "Ventas positivas": int(data["is_positive_sale"].sum()),
        "Facturación positiva": float(sales["TotalPrice"].sum()),
        "Unidades positivas": float(sales["Quantity"].sum()),
        "Fecha inicio": data["InvoiceDate"].min(),
        "Fecha fin": data["InvoiceDate"].max(),
        "Países únicos": int(data["Country"].nunique(dropna=True)),
        "Productos únicos": int(data["Description"].nunique(dropna=True)),
        "Clientes únicos": int(data["CustomerID"].nunique(dropna=True))
    }

summary_df = pd.DataFrame([build_summary(name, data) for name, data in datasets.items()])

print("=" * 100)
print("TABLA RESUMEN COMPARATIVA")
print("=" * 100)
display(summary_df)

# ------------------------------------------------------------
# 3) TABLA DE IMPACTO DE LA LIMPIEZA
# ------------------------------------------------------------
impacto = pd.DataFrame({
    "Métrica": [
        "Filas totales",
        "Duplicados exactos",
        "Quantity <= 0",
        "UnitPrice <= 0",
        "Nulos CustomerID",
        "Facturación positiva",
        "Unidades positivas",
        "Productos únicos",
        "Clientes únicos"
    ],
    "Original (df)": [
        len(raw),
        int(raw.duplicated().sum()),
        int((pd.to_numeric(raw["Quantity"], errors="coerce") <= 0).sum()),
        int((pd.to_numeric(raw["UnitPrice"], errors="coerce") <= 0).sum()),
        int(pd.to_numeric(raw["CustomerID"], errors="coerce").isna().sum()),
        float(raw.loc[raw["is_positive_sale"], "TotalPrice"].sum()),
        float(raw.loc[raw["is_positive_sale"], "Quantity"].sum()),
        int(raw["Description"].nunique(dropna=True)),
        int(raw["CustomerID"].nunique(dropna=True))
    ],
    "Limpio (dfc)": [
        len(clean),
        int(clean.duplicated().sum()),
        int((pd.to_numeric(clean["Quantity"], errors="coerce") <= 0).sum()),
        int((pd.to_numeric(clean["UnitPrice"], errors="coerce") <= 0).sum()),
        int(pd.to_numeric(clean["CustomerID"], errors="coerce").isna().sum()),
        float(clean.loc[clean["is_positive_sale"], "TotalPrice"].sum()),
        float(clean.loc[clean["is_positive_sale"], "Quantity"].sum()),
        int(clean["Description"].nunique(dropna=True)),
        int(clean["CustomerID"].nunique(dropna=True))
    ]
})

impacto["Diferencia absoluta"] = impacto["Limpio (dfc)"] - impacto["Original (df)"]

print("=" * 100)
print("IMPACTO DE LA LIMPIEZA")
print("=" * 100)
display(impacto)

# ------------------------------------------------------------
# 4) FILAS: ORIGINAL VS LIMPIO VS CLUSTERING
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(12, 6))
filas_comp = summary_df[["Dataset", "Filas"]].copy()
sns.barplot(data=filas_comp, x="Dataset", y="Filas", ax=ax)
ax.set_title("Comparación de número de filas por dataset")
ax.set_xlabel("Dataset")
ax.set_ylabel("Número de filas")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
for i, v in enumerate(filas_comp["Filas"]):
    ax.text(i, v, f"{int(v):,}".replace(",", "."), ha="center", va="bottom", fontsize=11)
save_show(fig, "01_filas_por_dataset.png")

# ------------------------------------------------------------
# 5) NULOS POR COLUMNA: ORIGINAL VS LIMPIO
# ------------------------------------------------------------
cols_base = ["InvoiceNo", "StockCode", "Description", "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
null_compare = pd.DataFrame({
    "Columna": cols_base,
    "Original (df)": [raw[c].isna().sum() for c in cols_base],
    "Limpio (dfc)": [clean[c].isna().sum() for c in cols_base]
}).melt(id_vars="Columna", var_name="Dataset", value_name="Nulos")

fig, ax = plt.subplots(figsize=(14, 7))
sns.barplot(data=null_compare, x="Columna", y="Nulos", hue="Dataset", ax=ax)
ax.set_title("Valores nulos por columna: original vs limpio")
ax.set_xlabel("Columnas")
ax.set_ylabel("Número de valores nulos")
ax.tick_params(axis="x", rotation=25)
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, "02_nulos_por_columna.png")

# ------------------------------------------------------------
# 6) DISTRIBUCIONES: HISTOGRAMAS COMPARATIVOS
#    Se recortan al percentil 99 para hacerlos legibles
# ------------------------------------------------------------
numeric_cols = ["Quantity", "UnitPrice", "TotalPrice"]

for col in numeric_cols:
    s_raw = pd.to_numeric(raw[col], errors="coerce").dropna()
    s_clean = pd.to_numeric(clean[col], errors="coerce").dropna()

    p01 = min(s_raw.quantile(0.01), s_clean.quantile(0.01))
    p99 = max(s_raw.quantile(0.99), s_clean.quantile(0.99))

    plot_df = pd.concat([
        pd.DataFrame({"Valor": s_raw.clip(lower=p01, upper=p99), "Dataset": "Original (df)"}),
        pd.DataFrame({"Valor": s_clean.clip(lower=p01, upper=p99), "Dataset": "Limpio (dfc)"})
    ], ignore_index=True)

    fig, ax = plt.subplots(figsize=(14, 7))
    sns.histplot(
        data=plot_df,
        x="Valor",
        hue="Dataset",
        bins=60,
        stat="density",
        common_norm=False,
        kde=True,
        ax=ax,
        alpha=0.35
    )
    ax.set_title(f"Distribución comparativa de {col} (recortada entre p1 y p99)")
    ax.set_xlabel(col)
    ax.set_ylabel("Densidad")
    save_show(fig, f"03_hist_{col.lower()}.png")

# ------------------------------------------------------------
# 7) BOXPLOTS COMPARATIVOS
#    También recortados para evitar que los extremos destruyan la escala
# ------------------------------------------------------------
for col in numeric_cols:
    s_raw = pd.to_numeric(raw[col], errors="coerce").dropna()
    s_clean = pd.to_numeric(clean[col], errors="coerce").dropna()

    p01 = min(s_raw.quantile(0.01), s_clean.quantile(0.01))
    p99 = max(s_raw.quantile(0.99), s_clean.quantile(0.99))

    box_df = pd.concat([
        pd.DataFrame({"Valor": s_raw.clip(lower=p01, upper=p99), "Dataset": "Original (df)"}),
        pd.DataFrame({"Valor": s_clean.clip(lower=p01, upper=p99), "Dataset": "Limpio (dfc)"})
    ], ignore_index=True)

    fig, ax = plt.subplots(figsize=(12, 6))
    sns.boxplot(data=box_df, x="Dataset", y="Valor", ax=ax)
    ax.set_title(f"Boxplot comparativo de {col} (recortado entre p1 y p99)")
    ax.set_xlabel("Dataset")
    ax.set_ylabel(col)
    save_show(fig, f"04_boxplot_{col.lower()}.png")

# ------------------------------------------------------------
# 8) % OUTLIERS POR IQR
# ------------------------------------------------------------
outlier_df = pd.DataFrame({
    "Variable": numeric_cols * 2,
    "Dataset": ["Original (df)"] * 3 + ["Limpio (dfc)"] * 3,
    "Porcentaje_outliers_IQR": [
        iqr_outlier_pct(raw["Quantity"]),
        iqr_outlier_pct(raw["UnitPrice"]),
        iqr_outlier_pct(raw["TotalPrice"]),
        iqr_outlier_pct(clean["Quantity"]),
        iqr_outlier_pct(clean["UnitPrice"]),
        iqr_outlier_pct(clean["TotalPrice"])
    ]
})

fig, ax = plt.subplots(figsize=(12, 6))
sns.barplot(data=outlier_df, x="Variable", y="Porcentaje_outliers_IQR", hue="Dataset", ax=ax)
ax.set_title("Porcentaje de outliers detectados por IQR")
ax.set_xlabel("Variable")
ax.set_ylabel("% de outliers")
for c in ax.containers:
    ax.bar_label(c, fmt="%.2f", fontsize=10)
save_show(fig, "05_outliers_iqr.png")

# ------------------------------------------------------------
# 9) FACTURACIÓN MENSUAL OBSERVADA
# ------------------------------------------------------------
def monthly_sales(data, label):
    tmp = data[data["is_positive_sale"]].copy()
    m = tmp.groupby("YearMonth", as_index=False)["TotalPrice"].sum()
    m["Dataset"] = label
    return m

monthly_compare = pd.concat([
    monthly_sales(raw, "Original (df)"),
    monthly_sales(clean, "Limpio (dfc)")
], ignore_index=True)

fig, ax = plt.subplots(figsize=(16, 7))
sns.lineplot(data=monthly_compare, x="YearMonth", y="TotalPrice", hue="Dataset", marker="o", linewidth=2.5, ax=ax)
ax.set_title("Facturación mensual observada: original vs limpio")
ax.set_xlabel("Mes")
ax.set_ylabel("Facturación positiva")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
ax.tick_params(axis="x", rotation=45)
ax.axvline(x="2011-12", linestyle="--", alpha=0.7)
save_show(fig, "06_facturacion_mensual_comparada.png")

# ------------------------------------------------------------
# 10) FACTURACIÓN SEMANAL OBSERVADA
# ------------------------------------------------------------
def weekly_sales(data, label):
    tmp = data[data["is_positive_sale"]].copy()
    s = tmp.set_index("InvoiceDate")["TotalPrice"].resample("W").sum().reset_index()
    s["Dataset"] = label
    return s

weekly_compare = pd.concat([
    weekly_sales(raw, "Original (df)"),
    weekly_sales(clean, "Limpio (dfc)")
], ignore_index=True)

fig, ax = plt.subplots(figsize=(16, 7))
sns.lineplot(data=weekly_compare, x="InvoiceDate", y="TotalPrice", hue="Dataset", linewidth=2, ax=ax)
ax.set_title("Facturación semanal observada: original vs limpio")
ax.set_xlabel("Fecha")
ax.set_ylabel("Facturación positiva semanal")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, "07_facturacion_semanal_comparada.png")

# ------------------------------------------------------------
# 11) COMPARACIÓN DICIEMBRE 2010 vs DICIEMBRE 2011 (DÍAS OBSERVADOS)
#     Solo en dfc, porque es el dataset final para modelado
# ------------------------------------------------------------
clean_sales = clean[clean["is_positive_sale"]].copy()
clean_sales["Year"] = clean_sales["InvoiceDate"].dt.year
clean_sales["Month"] = clean_sales["InvoiceDate"].dt.month
clean_sales["Day"] = clean_sales["InvoiceDate"].dt.day

dic_compare = clean_sales[
    (clean_sales["Month"] == 12) &
    (clean_sales["Year"].isin([2010, 2011])) &
    (clean_sales["Day"] <= 9)
].groupby(["Year", "Day"], as_index=False)["TotalPrice"].sum()

fig, ax = plt.subplots(figsize=(14, 7))
sns.lineplot(data=dic_compare, x="Day", y="TotalPrice", hue="Year", marker="o", linewidth=2.5, ax=ax)
ax.set_title("Comparación de facturación diaria: 1-9 diciembre 2010 vs 1-9 diciembre 2011 (dfc)")
ax.set_xlabel("Día de diciembre")
ax.set_ylabel("Facturación positiva")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, "08_comparacion_dic_1_9.png")

# ------------------------------------------------------------
# 12) TOP 10 PAÍSES POR NÚMERO DE FILAS
# ------------------------------------------------------------
top_countries_raw = raw["Country"].value_counts().head(10).rename_axis("Country").reset_index(name="Filas")
top_countries_raw["Dataset"] = "Original (df)"

top_countries_clean = clean["Country"].value_counts().head(10).rename_axis("Country").reset_index(name="Filas")
top_countries_clean["Dataset"] = "Limpio (dfc)"

country_compare = pd.concat([top_countries_raw, top_countries_clean], ignore_index=True)

fig, ax = plt.subplots(figsize=(16, 8))
sns.barplot(data=country_compare, y="Country", x="Filas", hue="Dataset", ax=ax)
ax.set_title("Top 10 países por número de filas")
ax.set_xlabel("Número de filas")
ax.set_ylabel("País")
ax.xaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, "09_top_paises.png")

# ------------------------------------------------------------
# 13) TOP 15 PRODUCTOS POR FACTURACIÓN EN dfc
# ------------------------------------------------------------
top_products_clean = (
    clean.loc[clean["is_positive_sale"]]
    .groupby("Description", as_index=False)["TotalPrice"]
    .sum()
    .sort_values("TotalPrice", ascending=False)
    .head(15)
)

fig, ax = plt.subplots(figsize=(16, 9))
sns.barplot(data=top_products_clean, y="Description", x="TotalPrice", ax=ax)
ax.set_title("Top 15 productos por facturación en el dataset limpio (dfc)")
ax.set_xlabel("Facturación positiva")
ax.set_ylabel("Descripción del producto")
ax.xaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, "10_top_productos_facturacion_dfc.png")

# ------------------------------------------------------------
# 14) TOP 15 PRODUCTOS POR UNIDADES EN dfc
# ------------------------------------------------------------
top_units_clean = (
    clean.loc[clean["is_positive_sale"]]
    .groupby("Description", as_index=False)["Quantity"]
    .sum()
    .sort_values("Quantity", ascending=False)
    .head(15)
)

fig, ax = plt.subplots(figsize=(16, 9))
sns.barplot(data=top_units_clean, y="Description", x="Quantity", ax=ax)
ax.set_title("Top 15 productos por unidades vendidas en el dataset limpio (dfc)")
ax.set_xlabel("Unidades vendidas")
ax.set_ylabel("Descripción del producto")
ax.xaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, "11_top_productos_unidades_dfc.png")

# ------------------------------------------------------------
# 15) DISTRIBUCIÓN DE FACTURAS POR NÚMERO DE LÍNEAS
# ------------------------------------------------------------
def invoice_lines_dist(data, label):
    tmp = data.groupby("InvoiceNo").size().reset_index(name="NumLineas")
    tmp["Dataset"] = label
    return tmp

invoice_lines = pd.concat([
    invoice_lines_dist(raw, "Original (df)"),
    invoice_lines_dist(clean, "Limpio (dfc)")
], ignore_index=True)

# Recorte para visualizar mejor
cap = invoice_lines["NumLineas"].quantile(0.99)
invoice_lines["NumLineas_plot"] = invoice_lines["NumLineas"].clip(upper=cap)

fig, ax = plt.subplots(figsize=(14, 7))
sns.histplot(data=invoice_lines, x="NumLineas_plot", hue="Dataset", bins=50, kde=True, stat="density", common_norm=False, ax=ax, alpha=0.35)
ax.set_title("Distribución del número de líneas por factura (recortada al p99)")
ax.set_xlabel("Número de líneas por factura")
ax.set_ylabel("Densidad")
save_show(fig, "12_lineas_por_factura.png")

# ------------------------------------------------------------
# 16) TABLA DE MESES OBSERVADOS Y ESTADO TEMPORAL
# ------------------------------------------------------------
monthly_status = (
    clean.groupby("YearMonth", as_index=False)
    .agg(
        FechaMin=("InvoiceDate", "min"),
        FechaMax=("InvoiceDate", "max"),
        Filas=("InvoiceNo", "count"),
        Facturacion=("TotalPrice", "sum")
    )
)

monthly_status["DiaInicioObservado"] = monthly_status["FechaMin"].dt.day
monthly_status["DiaFinObservado"] = monthly_status["FechaMax"].dt.day
monthly_status["MesEstado"] = "Completo / intermedio"
monthly_status.loc[monthly_status["YearMonth"] == monthly_status["YearMonth"].min(), "MesEstado"] = "Inicio del dataset"
monthly_status.loc[monthly_status["YearMonth"] == monthly_status["YearMonth"].max(), "MesEstado"] = "Final truncado"

print("=" * 100)
print("TABLA DE MESES OBSERVADOS EN dfc")
print("=" * 100)
display(monthly_status)

# ------------------------------------------------------------
# 17) TABLA FINAL DE RESULTADO
# ------------------------------------------------------------
resultado_final = pd.DataFrame({
    "Aspecto": [
        "Dataset original",
        "Dataset limpio",
        "Filas eliminadas",
        "% filas eliminadas",
        "Duplicados exactos restantes en dfc",
        "Quantity <= 0 restantes en dfc",
        "UnitPrice <= 0 restantes en dfc",
        "Facturación positiva final (dfc)",
        "Unidades positivas finales (dfc)",
        "Fecha inicio final",
        "Fecha fin final"
    ],
    "Valor": [
        len(raw),
        len(clean),
        len(raw) - len(clean),
        round(((len(raw) - len(clean)) / len(raw)) * 100, 2),
        int(clean.duplicated().sum()),
        int((clean["Quantity"] <= 0).sum()),
        int((clean["UnitPrice"] <= 0).sum()),
        round(clean.loc[clean["is_positive_sale"], "TotalPrice"].sum(), 2),
        int(clean.loc[clean["is_positive_sale"], "Quantity"].sum()),
        str(clean["InvoiceDate"].min()),
        str(clean["InvoiceDate"].max())
    ]
})

print("=" * 100)
print("RESULTADO FINAL DEL PREPROCESAMIENTO")
print("=" * 100)
display(resultado_final)

print(f"\nGráficos guardados en la carpeta: {OUTPUT_DIR}")
print("Celda final ejecutada correctamente.")
```

```python
# ============================================================
# CELDA MAESTRA FINAL - EDA COMPARATIVO COMPLETO
# df (original) vs dfc (limpio) [+ opcional dfc_clustering]
# Genera tablas, gráficos 2D, gráficos 3D y exportaciones
# ============================================================

import os
import math
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from matplotlib.ticker import FuncFormatter
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from IPython.display import display

# ------------------------------------------------------------
# 0) CONFIGURACIÓN GENERAL
# ------------------------------------------------------------
sns.set_theme(style="whitegrid", context="talk", palette="deep")

plt.rcParams["figure.figsize"] = (14, 7)
plt.rcParams["axes.titlesize"] = 18
plt.rcParams["axes.labelsize"] = 13
plt.rcParams["xtick.labelsize"] = 11
plt.rcParams["ytick.labelsize"] = 11

BASE_DIR = "EDA_FINAL_MASTER"
DIR_TABLES = os.path.join(BASE_DIR, "tablas")
DIR_2D = os.path.join(BASE_DIR, "graficos_2d")
DIR_3D = os.path.join(BASE_DIR, "graficos_3d")

for d in [BASE_DIR, DIR_TABLES, DIR_2D, DIR_3D]:
    os.makedirs(d, exist_ok=True)

def euro_fmt(x, pos):
    return f"{x:,.0f}".replace(",", ".")

def save_show(fig, path=None):
    if path:
        fig.savefig(path, dpi=180, bbox_inches="tight")
    plt.show()

def clip_series_for_plot(series, q_low=0.01, q_high=0.99):
    s = pd.to_numeric(series, errors="coerce").dropna()
    if len(s) == 0:
        return s
    low = s.quantile(q_low)
    high = s.quantile(q_high)
    return s.clip(lower=low, upper=high)

def iqr_outlier_pct(series):
    s = pd.to_numeric(series, errors="coerce").dropna()
    if len(s) < 5:
        return np.nan
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    return ((s < low) | (s > high)).mean() * 100

def preparar_dataset(data):
    d = data.copy()

    d["InvoiceDate"] = pd.to_datetime(d["InvoiceDate"], errors="coerce")
    d["Quantity"] = pd.to_numeric(d["Quantity"], errors="coerce")
    d["UnitPrice"] = pd.to_numeric(d["UnitPrice"], errors="coerce")
    d["CustomerID"] = pd.to_numeric(d.get("CustomerID"), errors="coerce")

    for col in ["InvoiceNo", "StockCode", "Description", "Country"]:
        if col in d.columns:
            d[col] = d[col].astype(str).str.strip()

    d["TotalPrice"] = d["Quantity"] * d["UnitPrice"]
    d["is_positive_sale"] = (d["Quantity"] > 0) & (d["UnitPrice"] > 0)
    d["is_return"] = (d["Quantity"] < 0) | (d["InvoiceNo"].astype(str).str.upper().str.startswith("C", na=False))

    d["Date"] = d["InvoiceDate"].dt.date
    d["YearMonth"] = d["InvoiceDate"].dt.to_period("M").astype(str)
    d["Year"] = d["InvoiceDate"].dt.year
    d["Month"] = d["InvoiceDate"].dt.month
    d["Day"] = d["InvoiceDate"].dt.day
    d["Weekday"] = d["InvoiceDate"].dt.dayofweek
    d["Hour"] = d["InvoiceDate"].dt.hour

    return d

# ------------------------------------------------------------
# 1) PREPARAR DATASETS
# ------------------------------------------------------------
raw = preparar_dataset(df)
clean = preparar_dataset(dfc)

datasets = {
    "Original (df)": raw,
    "Limpio (dfc)": clean
}

if "dfc_clustering" in globals():
    datasets["Clustering (dfc_clustering)"] = preparar_dataset(dfc_clustering)

# ------------------------------------------------------------
# 2) TABLAS RESUMEN
# ------------------------------------------------------------
def build_summary(name, data):
    sales = data[data["is_positive_sale"]].copy()

    description_nulls = data["Description"].replace("nan", np.nan).isna().sum() if "Description" in data.columns else np.nan

    return {
        "Dataset": name,
        "Filas": len(data),
        "Columnas": data.shape[1],
        "Duplicados exactos": int(data.drop(columns=["is_positive_sale", "is_return"], errors="ignore").duplicated().sum()),
        "Nulos Description": int(description_nulls),
        "Nulos CustomerID": int(data["CustomerID"].isna().sum()),
        "Quantity <= 0": int((data["Quantity"] <= 0).sum()),
        "UnitPrice <= 0": int((data["UnitPrice"] <= 0).sum()),
        "Filas devoluciones": int(data["is_return"].sum()),
        "Ventas positivas": int(data["is_positive_sale"].sum()),
        "Facturación positiva": float(sales["TotalPrice"].sum()),
        "Unidades positivas": float(sales["Quantity"].sum()),
        "Fecha inicio": data["InvoiceDate"].min(),
        "Fecha fin": data["InvoiceDate"].max(),
        "Países únicos": int(data["Country"].nunique(dropna=True)),
        "Productos únicos": int(data["Description"].nunique(dropna=True)),
        "Clientes únicos": int(data["CustomerID"].nunique(dropna=True))
    }

summary_df = pd.DataFrame([build_summary(name, data) for name, data in datasets.items()])
summary_df.to_csv(os.path.join(DIR_TABLES, "01_tabla_resumen_comparativa.csv"), index=False)

impacto = pd.DataFrame({
    "Métrica": [
        "Filas totales",
        "Duplicados exactos",
        "Quantity <= 0",
        "UnitPrice <= 0",
        "Nulos CustomerID",
        "Facturación positiva",
        "Unidades positivas",
        "Productos únicos",
        "Clientes únicos"
    ],
    "Original (df)": [
        len(raw),
        int(raw.duplicated().sum()),
        int((raw["Quantity"] <= 0).sum()),
        int((raw["UnitPrice"] <= 0).sum()),
        int(raw["CustomerID"].isna().sum()),
        float(raw.loc[raw["is_positive_sale"], "TotalPrice"].sum()),
        float(raw.loc[raw["is_positive_sale"], "Quantity"].sum()),
        int(raw["Description"].nunique(dropna=True)),
        int(raw["CustomerID"].nunique(dropna=True))
    ],
    "Limpio (dfc)": [
        len(clean),
        int(clean.duplicated().sum()),
        int((clean["Quantity"] <= 0).sum()),
        int((clean["UnitPrice"] <= 0).sum()),
        int(clean["CustomerID"].isna().sum()),
        float(clean.loc[clean["is_positive_sale"], "TotalPrice"].sum()),
        float(clean.loc[clean["is_positive_sale"], "Quantity"].sum()),
        int(clean["Description"].nunique(dropna=True)),
        int(clean["CustomerID"].nunique(dropna=True))
    ]
})
impacto["Diferencia absoluta"] = impacto["Limpio (dfc)"] - impacto["Original (df)"]
impacto.to_csv(os.path.join(DIR_TABLES, "02_impacto_limpieza.csv"), index=False)

monthly_status = (
    clean.groupby("YearMonth", as_index=False)
    .agg(
        FechaMin=("InvoiceDate", "min"),
        FechaMax=("InvoiceDate", "max"),
        Filas=("InvoiceNo", "count"),
        Facturacion=("TotalPrice", "sum")
    )
)
monthly_status["DiaInicioObservado"] = monthly_status["FechaMin"].dt.day
monthly_status["DiaFinObservado"] = monthly_status["FechaMax"].dt.day
monthly_status["MesEstado"] = "Completo / intermedio"
monthly_status.loc[monthly_status["YearMonth"] == monthly_status["YearMonth"].min(), "MesEstado"] = "Inicio del dataset"
monthly_status.loc[monthly_status["YearMonth"] == monthly_status["YearMonth"].max(), "MesEstado"] = "Final truncado"
monthly_status.to_csv(os.path.join(DIR_TABLES, "03_estado_meses_observados_dfc.csv"), index=False)

resultado_final = pd.DataFrame({
    "Aspecto": [
        "Dataset original",
        "Dataset limpio",
        "Filas eliminadas",
        "% filas eliminadas",
        "Duplicados exactos restantes en dfc",
        "Quantity <= 0 restantes en dfc",
        "UnitPrice <= 0 restantes en dfc",
        "Facturación positiva final (dfc)",
        "Unidades positivas finales (dfc)",
        "Fecha inicio final",
        "Fecha fin final"
    ],
    "Valor": [
        len(raw),
        len(clean),
        len(raw) - len(clean),
        round(((len(raw) - len(clean)) / len(raw)) * 100, 2),
        int(clean.duplicated().sum()),
        int((clean["Quantity"] <= 0).sum()),
        int((clean["UnitPrice"] <= 0).sum()),
        round(clean.loc[clean["is_positive_sale"], "TotalPrice"].sum(), 2),
        int(clean.loc[clean["is_positive_sale"], "Quantity"].sum()),
        str(clean["InvoiceDate"].min()),
        str(clean["InvoiceDate"].max())
    ]
})
resultado_final.to_csv(os.path.join(DIR_TABLES, "04_resultado_final_preprocesamiento.csv"), index=False)

print("=" * 100)
print("TABLAS GENERADAS")
print("=" * 100)
display(summary_df)
display(impacto)
display(monthly_status.head())
display(resultado_final)

# ------------------------------------------------------------
# 3) GRÁFICOS 2D
# ------------------------------------------------------------

# 3.1 Filas por dataset
fig, ax = plt.subplots(figsize=(12, 6))
tmp = summary_df[["Dataset", "Filas"]].copy()
sns.barplot(data=tmp, x="Dataset", y="Filas", ax=ax)
ax.set_title("Comparación del número de filas por dataset")
ax.set_xlabel("Dataset")
ax.set_ylabel("Número de filas")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
for i, v in enumerate(tmp["Filas"]):
    ax.text(i, v, f"{int(v):,}".replace(",", "."), ha="center", va="bottom", fontsize=11)
save_show(fig, os.path.join(DIR_2D, "01_filas_por_dataset.png"))

# 3.2 Nulos por columna
cols_base = ["InvoiceNo", "StockCode", "Description", "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
null_compare = pd.DataFrame({
    "Columna": cols_base,
    "Original (df)": [raw[c].isna().sum() for c in cols_base],
    "Limpio (dfc)": [clean[c].isna().sum() for c in cols_base]
}).melt(id_vars="Columna", var_name="Dataset", value_name="Nulos")

fig, ax = plt.subplots(figsize=(14, 7))
sns.barplot(data=null_compare, x="Columna", y="Nulos", hue="Dataset", ax=ax)
ax.set_title("Valores nulos por columna: original vs limpio")
ax.set_xlabel("Columna")
ax.set_ylabel("Número de nulos")
ax.tick_params(axis="x", rotation=25)
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, os.path.join(DIR_2D, "02_nulos_por_columna.png"))

# 3.3 Histogramas comparativos
numeric_cols = ["Quantity", "UnitPrice", "TotalPrice"]
for col in numeric_cols:
    s_raw = clip_series_for_plot(raw[col], 0.01, 0.99)
    s_clean = clip_series_for_plot(clean[col], 0.01, 0.99)

    plot_df = pd.concat([
        pd.DataFrame({"Valor": s_raw, "Dataset": "Original (df)"}),
        pd.DataFrame({"Valor": s_clean, "Dataset": "Limpio (dfc)"})
    ], ignore_index=True)

    fig, ax = plt.subplots(figsize=(14, 7))
    sns.histplot(
        data=plot_df,
        x="Valor",
        hue="Dataset",
        bins=60,
        stat="density",
        common_norm=False,
        kde=True,
        alpha=0.35,
        ax=ax
    )
    ax.set_title(f"Distribución comparativa de {col} (recorte p1-p99)")
    ax.set_xlabel(col)
    ax.set_ylabel("Densidad")
    save_show(fig, os.path.join(DIR_2D, f"03_hist_{col.lower()}.png"))

# 3.4 Boxplots comparativos
for col in numeric_cols:
    s_raw = clip_series_for_plot(raw[col], 0.01, 0.99)
    s_clean = clip_series_for_plot(clean[col], 0.01, 0.99)

    box_df = pd.concat([
        pd.DataFrame({"Valor": s_raw, "Dataset": "Original (df)"}),
        pd.DataFrame({"Valor": s_clean, "Dataset": "Limpio (dfc)"})
    ], ignore_index=True)

    fig, ax = plt.subplots(figsize=(12, 6))
    sns.boxplot(data=box_df, x="Dataset", y="Valor", ax=ax)
    ax.set_title(f"Boxplot comparativo de {col} (recorte p1-p99)")
    ax.set_xlabel("Dataset")
    ax.set_ylabel(col)
    save_show(fig, os.path.join(DIR_2D, f"04_boxplot_{col.lower()}.png"))

# 3.5 Porcentaje de outliers por IQR
outlier_df = pd.DataFrame({
    "Variable": numeric_cols * 2,
    "Dataset": ["Original (df)"] * 3 + ["Limpio (dfc)"] * 3,
    "Porcentaje_outliers_IQR": [
        iqr_outlier_pct(raw["Quantity"]),
        iqr_outlier_pct(raw["UnitPrice"]),
        iqr_outlier_pct(raw["TotalPrice"]),
        iqr_outlier_pct(clean["Quantity"]),
        iqr_outlier_pct(clean["UnitPrice"]),
        iqr_outlier_pct(clean["TotalPrice"])
    ]
})

fig, ax = plt.subplots(figsize=(12, 6))
sns.barplot(data=outlier_df, x="Variable", y="Porcentaje_outliers_IQR", hue="Dataset", ax=ax)
ax.set_title("Porcentaje de outliers detectados por IQR")
ax.set_xlabel("Variable")
ax.set_ylabel("% de outliers")
for c in ax.containers:
    ax.bar_label(c, fmt="%.2f", fontsize=10)
save_show(fig, os.path.join(DIR_2D, "05_outliers_iqr.png"))

# 3.6 Facturación mensual
def monthly_sales(data, label):
    tmp = data[data["is_positive_sale"]].copy()
    m = tmp.groupby("YearMonth", as_index=False)["TotalPrice"].sum()
    m["Dataset"] = label
    return m

monthly_compare = pd.concat([
    monthly_sales(raw, "Original (df)"),
    monthly_sales(clean, "Limpio (dfc)")
], ignore_index=True)

fig, ax = plt.subplots(figsize=(16, 7))
sns.lineplot(data=monthly_compare, x="YearMonth", y="TotalPrice", hue="Dataset", marker="o", linewidth=2.5, ax=ax)
ax.set_title("Facturación mensual observada: original vs limpio")
ax.set_xlabel("Mes")
ax.set_ylabel("Facturación positiva")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
ax.tick_params(axis="x", rotation=45)
save_show(fig, os.path.join(DIR_2D, "06_facturacion_mensual_comparada.png"))

# 3.7 Facturación semanal
def weekly_sales(data, label):
    tmp = data[data["is_positive_sale"]].copy()
    s = tmp.set_index("InvoiceDate")["TotalPrice"].resample("W").sum().reset_index()
    s["Dataset"] = label
    return s

weekly_compare = pd.concat([
    weekly_sales(raw, "Original (df)"),
    weekly_sales(clean, "Limpio (dfc)")
], ignore_index=True)

fig, ax = plt.subplots(figsize=(16, 7))
sns.lineplot(data=weekly_compare, x="InvoiceDate", y="TotalPrice", hue="Dataset", linewidth=2, ax=ax)
ax.set_title("Facturación semanal observada: original vs limpio")
ax.set_xlabel("Fecha")
ax.set_ylabel("Facturación semanal positiva")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, os.path.join(DIR_2D, "07_facturacion_semanal_comparada.png"))

# 3.8 Comparación 1-9 diciembre 2010 vs 2011
clean_sales = clean[clean["is_positive_sale"]].copy()
dic_compare = clean_sales[
    (clean_sales["Month"] == 12) &
    (clean_sales["Year"].isin([2010, 2011])) &
    (clean_sales["Day"] <= 9)
].groupby(["Year", "Day"], as_index=False)["TotalPrice"].sum()

fig, ax = plt.subplots(figsize=(14, 7))
sns.lineplot(data=dic_compare, x="Day", y="TotalPrice", hue="Year", marker="o", linewidth=2.5, ax=ax)
ax.set_title("Comparación de facturación diaria: 1-9 diciembre 2010 vs 1-9 diciembre 2011 (dfc)")
ax.set_xlabel("Día de diciembre")
ax.set_ylabel("Facturación positiva")
ax.yaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, os.path.join(DIR_2D, "08_comparacion_dic_1_9.png"))

# 3.9 Top países
top_countries_raw = raw["Country"].value_counts().head(10).rename_axis("Country").reset_index(name="Filas")
top_countries_raw["Dataset"] = "Original (df)"
top_countries_clean = clean["Country"].value_counts().head(10).rename_axis("Country").reset_index(name="Filas")
top_countries_clean["Dataset"] = "Limpio (dfc)"
country_compare = pd.concat([top_countries_raw, top_countries_clean], ignore_index=True)

fig, ax = plt.subplots(figsize=(16, 8))
sns.barplot(data=country_compare, y="Country", x="Filas", hue="Dataset", ax=ax)
ax.set_title("Top 10 países por número de filas")
ax.set_xlabel("Número de filas")
ax.set_ylabel("País")
ax.xaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, os.path.join(DIR_2D, "09_top_paises.png"))

# 3.10 Top productos por facturación en dfc
top_products_clean = (
    clean.loc[clean["is_positive_sale"]]
    .groupby("Description", as_index=False)["TotalPrice"]
    .sum()
    .sort_values("TotalPrice", ascending=False)
    .head(15)
)

fig, ax = plt.subplots(figsize=(16, 9))
sns.barplot(data=top_products_clean, y="Description", x="TotalPrice", ax=ax)
ax.set_title("Top 15 productos por facturación en el dataset limpio (dfc)")
ax.set_xlabel("Facturación positiva")
ax.set_ylabel("Descripción del producto")
ax.xaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, os.path.join(DIR_2D, "10_top_productos_facturacion_dfc.png"))

# 3.11 Top productos por unidades
top_units_clean = (
    clean.loc[clean["is_positive_sale"]]
    .groupby("Description", as_index=False)["Quantity"]
    .sum()
    .sort_values("Quantity", ascending=False)
    .head(15)
)

fig, ax = plt.subplots(figsize=(16, 9))
sns.barplot(data=top_units_clean, y="Description", x="Quantity", ax=ax)
ax.set_title("Top 15 productos por unidades vendidas en el dataset limpio (dfc)")
ax.set_xlabel("Unidades vendidas")
ax.set_ylabel("Descripción del producto")
ax.xaxis.set_major_formatter(FuncFormatter(euro_fmt))
save_show(fig, os.path.join(DIR_2D, "11_top_productos_unidades_dfc.png"))

# 3.12 Distribución de líneas por factura
def invoice_lines_dist(data, label):
    tmp = data.groupby("InvoiceNo").size().reset_index(name="NumLineas")
    tmp["Dataset"] = label
    return tmp

invoice_lines = pd.concat([
    invoice_lines_dist(raw, "Original (df)"),
    invoice_lines_dist(clean, "Limpio (dfc)")
], ignore_index=True)

cap = invoice_lines["NumLineas"].quantile(0.99)
invoice_lines["NumLineas_plot"] = invoice_lines["NumLineas"].clip(upper=cap)

fig, ax = plt.subplots(figsize=(14, 7))
sns.histplot(data=invoice_lines, x="NumLineas_plot", hue="Dataset", bins=50, kde=True, stat="density", common_norm=False, ax=ax, alpha=0.35)
ax.set_title("Distribución del número de líneas por factura (recorte al p99)")
ax.set_xlabel("Número de líneas por factura")
ax.set_ylabel("Densidad")
save_show(fig, os.path.join(DIR_2D, "12_lineas_por_factura.png"))

# ------------------------------------------------------------
# 4) GRÁFICOS 3D
# ------------------------------------------------------------

# 4.1 Scatter 3D
def sample_for_3d_scatter(data, max_points=4000):
    s = data[data["is_positive_sale"]].copy()
    s = s[["Quantity", "UnitPrice", "TotalPrice"]].dropna()
    for col in ["Quantity", "UnitPrice", "TotalPrice"]:
        p1 = s[col].quantile(0.01)
        p99 = s[col].quantile(0.99)
        s = s[(s[col] >= p1) & (s[col] <= p99)]
    if len(s) > max_points:
        s = s.sample(max_points, random_state=42)
    return s

scatter_raw = sample_for_3d_scatter(raw)
scatter_clean = sample_for_3d_scatter(clean)

fig = plt.figure(figsize=(18, 8))

ax1 = fig.add_subplot(121, projection="3d")
p1 = ax1.scatter(
    scatter_raw["Quantity"],
    scatter_raw["UnitPrice"],
    scatter_raw["TotalPrice"],
    c=scatter_raw["TotalPrice"],
    cmap="viridis",
    alpha=0.55,
    s=18
)
ax1.set_title("Scatter 3D - Dataset original (df)")
ax1.set_xlabel("Quantity")
ax1.set_ylabel("UnitPrice")
ax1.set_zlabel("TotalPrice")
fig.colorbar(p1, ax=ax1, shrink=0.6, pad=0.08, label="TotalPrice")

ax2 = fig.add_subplot(122, projection="3d")
p2 = ax2.scatter(
    scatter_clean["Quantity"],
    scatter_clean["UnitPrice"],
    scatter_clean["TotalPrice"],
    c=scatter_clean["TotalPrice"],
    cmap="plasma",
    alpha=0.55,
    s=18
)
ax2.set_title("Scatter 3D - Dataset limpio (dfc)")
ax2.set_xlabel("Quantity")
ax2.set_ylabel("UnitPrice")
ax2.set_zlabel("TotalPrice")
fig.colorbar(p2, ax=ax2, shrink=0.6, pad=0.08, label="TotalPrice")

plt.suptitle("Comparativa 3D de Quantity, UnitPrice y TotalPrice", y=1.02, fontsize=18)
plt.tight_layout()
save_show(fig, os.path.join(DIR_3D, "01_scatter3d_df_vs_dfc.png"))

# 4.2 Barras 3D mensuales
monthly_raw = (
    raw[raw["is_positive_sale"]]
    .groupby("YearMonth", as_index=False)["TotalPrice"]
    .sum()
)
monthly_raw["Dataset"] = "Original"

monthly_clean = (
    clean[clean["is_positive_sale"]]
    .groupby("YearMonth", as_index=False)["TotalPrice"]
    .sum()
)
monthly_clean["Dataset"] = "Limpio"

monthly_3d = pd.concat([monthly_raw, monthly_clean], ignore_index=True).sort_values(["YearMonth", "Dataset"])
months = sorted(monthly_3d["YearMonth"].unique().tolist())
month_map = {m: i for i, m in enumerate(months)}
dataset_map = {"Original": 0, "Limpio": 1}

monthly_3d["x"] = monthly_3d["YearMonth"].map(month_map)
monthly_3d["y"] = monthly_3d["Dataset"].map(dataset_map)

fig = plt.figure(figsize=(18, 9))
ax = fig.add_subplot(111, projection="3d")

dx = np.full(len(monthly_3d), 0.45)
dy = np.full(len(monthly_3d), 0.45)
dz = monthly_3d["TotalPrice"].values
colors = ["#4C72B0" if d == "Original" else "#55A868" for d in monthly_3d["Dataset"]]

ax.bar3d(
    monthly_3d["x"].values,
    monthly_3d["y"].values,
    np.zeros(len(monthly_3d)),
    dx, dy, dz,
    color=colors,
    alpha=0.85,
    shade=True
)

ax.set_title("Facturación mensual observada en 3D: df vs dfc")
ax.set_xlabel("Mes observado")
ax.set_ylabel("Dataset")
ax.set_zlabel("Facturación positiva")
ax.set_xticks(range(len(months)))
ax.set_xticklabels(months, rotation=45, ha="right")
ax.set_yticks([0, 1])
ax.set_yticklabels(["Original", "Limpio"])
save_show(fig, os.path.join(DIR_3D, "02_bar3d_facturacion_mensual.png"))

# 4.3 Superficie 3D día-semana x hora
weekday_labels = {0: "Lun", 1: "Mar", 2: "Mié", 3: "Jue", 4: "Vie", 5: "Sáb", 6: "Dom"}

surface_df = (
    clean[clean["is_positive_sale"]]
    .dropna(subset=["Weekday", "Hour", "TotalPrice"])
    .groupby(["Weekday", "Hour"], as_index=False)["TotalPrice"]
    .sum()
)

pivot_surface = surface_df.pivot(index="Weekday", columns="Hour", values="TotalPrice").fillna(0)
pivot_surface = pivot_surface.reindex(index=range(7), columns=range(24), fill_value=0)

X, Y = np.meshgrid(pivot_surface.columns.values, pivot_surface.index.values)
Z = pivot_surface.values

fig = plt.figure(figsize=(18, 9))
ax = fig.add_subplot(111, projection="3d")
surf = ax.plot_surface(
    X, Y, Z,
    cmap="coolwarm",
    edgecolor="none",
    alpha=0.95
)
ax.set_title("Superficie 3D de facturación en dfc por día de la semana y hora")
ax.set_xlabel("Hora del día")
ax.set_ylabel("Día de la semana")
ax.set_zlabel("Facturación positiva")
ax.set_yticks(range(7))
ax.set_yticklabels([weekday_labels[i] for i in range(7)])
fig.colorbar(surf, ax=ax, shrink=0.6, pad=0.08, label="Facturación")
save_show(fig, os.path.join(DIR_3D, "03_surface3d_weekday_hour_facturacion_dfc.png"))

# ------------------------------------------------------------
# 5) CONTACT SHEET OPCIONAL DE GRÁFICOS
# ------------------------------------------------------------
graficos_generados = (
    [os.path.join(DIR_2D, f) for f in sorted(os.listdir(DIR_2D)) if f.lower().endswith(".png")] +
    [os.path.join(DIR_3D, f) for f in sorted(os.listdir(DIR_3D)) if f.lower().endswith(".png")]
)

contact_rows = []
for i, path in enumerate(graficos_generados, start=1):
    contact_rows.append({
        "N": i,
        "Archivo": os.path.basename(path),
        "Ruta": path
    })

contact_df = pd.DataFrame(contact_rows)
contact_df.to_csv(os.path.join(DIR_TABLES, "05_indice_graficos.csv"), index=False)

# ------------------------------------------------------------
# 6) MENSAJE FINAL
# ------------------------------------------------------------
print("=" * 100)
print("EXPORTACIÓN COMPLETA FINALIZADA")
print("=" * 100)
print(f"Carpeta base: {BASE_DIR}")
print(f"Tablas CSV: {DIR_TABLES}")
print(f"Gráficos 2D: {DIR_2D}")
print(f"Gráficos 3D: {DIR_3D}")
print(f"Número total de gráficos generados: {len(graficos_generados)}")

print("\nArchivos CSV creados:")
for f in sorted(os.listdir(DIR_TABLES)):
    print("-", f)

print("\nGráficos 2D creados:")
for f in sorted(os.listdir(DIR_2D)):
    print("-", f)

print("\nGráficos 3D creados:")
for f in sorted(os.listdir(DIR_3D)):
    print("-", f)

print("\nCelda maestra ejecutada correctamente.")
```
