```python
# ============================================================
# CELDA 3 - DEFINIR RUTA DEL DATASET
# ============================================================

# Ajusta esta ruta a tu carpeta real de Drive
ruta_dataset = "/content/drive/MyDrive/PROYECTO TFM IA_BG_E1/CLUSTERING/datasets/data_v01_clustering_ventas.csv"

# Comprobamos si el archivo existe
if os.path.exists(ruta_dataset):
    print("Archivo encontrado correctamente.")
else:
    print("ERROR: No se encontró el archivo. Revisa la ruta.")
```

```python
# ============================================================
# CELDA 4 - CARGAR DATASET
# ============================================================

df = pd.read_csv(ruta_dataset, encoding="ISO-8859-1")

print("Dataset cargado correctamente.")
print("Filas y columnas:", df.shape)

display(df.head())
```

```python
# ============================================================
# CELDA 5 - EXPLORACIÓN INICIAL
# ============================================================

print("Información general del dataset:")
df.info()

print("\nPrimeras filas:")
display(df.head())

print("\nColumnas disponibles:")
print(df.columns.tolist())
```

```python
# ============================================================
# CELDA 6 - COMPROBAR NULOS INICIALES
# ============================================================

nulos = df.isnull().sum().sort_values(ascending=False)

print("Valores nulos por columna:")
display(nulos[nulos > 0])

print("\nPorcentaje de nulos por columna:")
porcentaje_nulos = (df.isnull().mean() * 100).round(2).sort_values(ascending=False)
display(porcentaje_nulos[porcentaje_nulos > 0])
```

```python
# ============================================================
# CELDA 7 - GUARDAR REFERENCIA INICIAL
# ============================================================

filas_iniciales = df.shape[0]
columnas_iniciales = df.shape[1]

print("Filas iniciales:", filas_iniciales)
print("Columnas iniciales:", columnas_iniciales)
```

```python
# ============================================================
# CELDA 8 - ELIMINAR CUSTOMERID NULOS
# ============================================================

filas_antes = df.shape[0]

df = df.dropna(subset=["CustomerID"]).copy()

filas_despues = df.shape[0]
filas_eliminadas_customer = filas_antes - filas_despues

print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas por CustomerID nulo:", filas_eliminadas_customer)
print("CustomerID nulos restantes:", df["CustomerID"].isnull().sum())
```

```python
# ============================================================
# CELDA 9 - ELIMINAR DESCRIPTION NULAS
# ============================================================

filas_antes = df.shape[0]

df = df.dropna(subset=["Description"]).copy()

filas_despues = df.shape[0]
filas_eliminadas_description = filas_antes - filas_despues

print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas por Description nula:", filas_eliminadas_description)
print("Description nulas restantes:", df["Description"].isnull().sum())
```

```python
# ============================================================
# CELDA 10 - ELIMINAR FACTURAS CANCELADAS
# ============================================================

# Las facturas canceladas suelen empezar por la letra C
facturas_canceladas = df["InvoiceNo"].astype(str).str.startswith("C").sum()

print("Facturas canceladas detectadas:", facturas_canceladas)

filas_antes = df.shape[0]

df = df[~df["InvoiceNo"].astype(str).str.startswith("C")].copy()

filas_despues = df.shape[0]
filas_eliminadas_canceladas = filas_antes - filas_despues

print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas por InvoiceNo cancelado:", filas_eliminadas_canceladas)
print("Facturas canceladas restantes:", df["InvoiceNo"].astype(str).str.startswith("C").sum())
```

```python
# ============================================================
# CELDA 11 - ELIMINAR QUANTITY <= 0
# ============================================================

# Quantity negativa indica devolución.
# Quantity igual a 0 no representa una compra real.

cantidad_invalida = (df["Quantity"] <= 0).sum()

print("Filas con Quantity <= 0 detectadas:", cantidad_invalida)

filas_antes = df.shape[0]

df = df[df["Quantity"] > 0].copy()

filas_despues = df.shape[0]
filas_eliminadas_quantity = filas_antes - filas_despues

print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas por Quantity <= 0:", filas_eliminadas_quantity)
print("Quantity <= 0 restantes:", (df["Quantity"] <= 0).sum())
```

```python
# ============================================================
# CELDA 12 - ELIMINAR UNITPRICE <= 0
# ============================================================

# UnitPrice igual o menor que 0 no representa una venta económica válida.
# Esta limpieza evita distorsionar TotalPrice, Monetary y AverageOrderValue.

precio_invalido = (df["UnitPrice"] <= 0).sum()

print("Filas con UnitPrice <= 0 detectadas:", precio_invalido)

filas_antes = df.shape[0]

df = df[df["UnitPrice"] > 0].copy()

filas_despues = df.shape[0]
filas_eliminadas_unitprice = filas_antes - filas_despues

print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas por UnitPrice <= 0:", filas_eliminadas_unitprice)
print("UnitPrice <= 0 restantes:", (df["UnitPrice"] <= 0).sum())
```

```python
# ============================================================
# CELDA 13 - CONVERTIR TIPOS DE DATOS
# ============================================================

# CustomerID es un identificador, no una variable numérica predictiva.
# Se convierte a string para evitar formatos como 17850.0.

df["CustomerID"] = df["CustomerID"].astype(int).astype(str)

# InvoiceDate debe estar en formato datetime para el feature engineering posterior.
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")

# Comprobación de fechas inválidas tras la conversión
fechas_invalidas = df["InvoiceDate"].isnull().sum()

print("Tipos convertidos correctamente.")
print("Fechas inválidas tras conversión:", fechas_invalidas)

print("\nTipo de CustomerID:", df["CustomerID"].dtype)
print("Tipo de InvoiceDate:", df["InvoiceDate"].dtype)

display(df[["CustomerID", "InvoiceDate"]].head())
```

```python
# ============================================================
# CELDA 14 - RECALCULAR TOTALPRICE Y VALIDAR IMPORTES
# ============================================================

# Recalculamos TotalPrice para asegurar coherencia después de la limpieza.
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

# Validamos importes nulos, negativos o cero.
totalprice_invalidos = (df["TotalPrice"] <= 0).sum()

print("Columna TotalPrice recalculada correctamente.")
print("Filas con TotalPrice <= 0 detectadas:", totalprice_invalidos)

# Si existieran importes no positivos, se eliminan.
filas_antes = df.shape[0]

df = df[df["TotalPrice"] > 0].copy()

filas_despues = df.shape[0]
filas_eliminadas_totalprice = filas_antes - filas_despues

print("Filas antes:", filas_antes)
print("Filas después:", filas_despues)
print("Filas eliminadas por TotalPrice <= 0:", filas_eliminadas_totalprice)
print("TotalPrice <= 0 restantes:", (df["TotalPrice"] <= 0).sum())

display(df[["Quantity", "UnitPrice", "TotalPrice"]].head())
```

```python
# ============================================================
# CELDA 15 - ELIMINAR COLUMNAS HEREDADAS DE REGRESIÓN
# ============================================================

columnas_regresion = [
    "Sales_Lag_1",
    "Sales_Lag_7",
    "Sales_Mean_7",
    "Sales_Mean_30",
    "sales_lag_1",
    "sales_lag_7",
    "rolling_mean_7",
    "rolling_mean_30",
    "DailyTotal",
    "daily_total",
    "Trend_Lag1_vs_Lag7",
    "trend_1_vs_7",
    "IsUK",
    "IsUk",
    "isUK",
    "is_UK",
    "DayOfWeek",
    "dayOfweek",
    "day_of_week",
    "IsWeekend",
    "is_weekend",
    "InvoiceDay",
    "InvoiceMonth",
    "InvoiceYear",
    "InvoiceHour"
]

columnas_a_borrar = [col for col in columnas_regresion if col in df.columns]

print("Columnas de regresión encontradas para borrar:")
print(columnas_a_borrar)

df = df.drop(columns=columnas_a_borrar)

print("\nColumnas eliminadas:", len(columnas_a_borrar))
print("Columnas restantes:", df.columns.tolist())
```

```python
# ============================================================
# CELDA 16 - ORDENAR COLUMNAS Y COMPROBACIÓN FINAL
# ============================================================

columnas_finales = [
    "InvoiceNo",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "CustomerID",
    "Country",
    "TotalPrice"
]

# Verificamos que todas las columnas necesarias existen.
columnas_faltantes = [col for col in columnas_finales if col not in df.columns]

if columnas_faltantes:
    raise ValueError(f"Faltan columnas necesarias en el dataset: {columnas_faltantes}")

# Dejamos el dataset solo con las columnas finales necesarias.
df = df[columnas_finales].copy()

print("COMPROBACIÓN FINAL DEL DATASET LIMPIO")
print("-" * 60)

print("Filas finales:", df.shape[0])
print("Columnas finales:", df.shape[1])

print("\nNulos en CustomerID:", df["CustomerID"].isnull().sum())
print("Nulos en Description:", df["Description"].isnull().sum())
print("Nulos en InvoiceDate:", df["InvoiceDate"].isnull().sum())

print("\nQuantity <= 0:", (df["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (df["UnitPrice"] <= 0).sum())
print("TotalPrice <= 0:", (df["TotalPrice"] <= 0).sum())

print("\nFacturas canceladas restantes:", df["InvoiceNo"].astype(str).str.startswith("C").sum())

print("\nClientes únicos:", df["CustomerID"].nunique())
print("Facturas únicas:", df["InvoiceNo"].nunique())
print("Productos únicos:", df["StockCode"].nunique())
print("Países presentes:", df["Country"].nunique())

print("\nPeriodo temporal:")
print("Fecha mínima:", df["InvoiceDate"].min())
print("Fecha máxima:", df["InvoiceDate"].max())

print("\nTotalPrice:")
print("Media:", round(df["TotalPrice"].mean(), 2))
print("Mínimo real:", df["TotalPrice"].min())
print("Máximo:", round(df["TotalPrice"].max(), 2))

print("\nColumnas finales:")
print(df.columns.tolist())
```

```python
# ============================================================
# CELDA 17 - RESUMEN DE LIMPIEZA
# ============================================================

filas_finales = df.shape[0]
filas_eliminadas_totales = filas_iniciales - filas_finales

resumen_limpieza = pd.DataFrame({
    "Paso": [
        "Dataset inicial",
        "CustomerID nulos eliminados",
        "Description nulas eliminadas",
        "Invoices canceladas eliminadas",
        "Quantity <= 0 eliminadas",
        "UnitPrice <= 0 eliminadas",
        "TotalPrice <= 0 eliminadas",
        "Dataset final limpio"
    ],
    "Filas": [
        filas_iniciales,
        filas_eliminadas_customer,
        filas_eliminadas_description,
        filas_eliminadas_canceladas,
        filas_eliminadas_quantity,
        filas_eliminadas_unitprice,
        filas_eliminadas_totalprice,
        filas_finales
    ]
})

resumen_limpieza["Porcentaje sobre inicial"] = (
    resumen_limpieza["Filas"] / filas_iniciales * 100
).round(2)

display(resumen_limpieza)

print("Filas eliminadas totales:", filas_eliminadas_totales)
print("Porcentaje eliminado:", round(filas_eliminadas_totales / filas_iniciales * 100, 2), "%")
print("Porcentaje conservado:", round(filas_finales / filas_iniciales * 100, 2), "%")
```

```python
# ============================================================
# CELDA 18 - VISTA FINAL DEL DATASET LIMPIO
# ============================================================

display(df.head())

print("Dimensiones finales:", df.shape)

print("\nColumnas finales:")
print(df.columns.tolist())

print("\nTipos de datos finales:")
display(df.dtypes)
```

```python
# ============================================================
# CELDA 27 - BOXPLOTS INTERACTIVOS DE VARIABLES ECONÓMICAS
# ============================================================

df_box = df[["Quantity", "UnitPrice", "TotalPrice"]].copy()

fig = px.box(
    df_box,
    y=["Quantity", "UnitPrice", "TotalPrice"],
    title="Distribución y valores extremos tras la limpieza"
)

fig.update_layout(
    yaxis_title="Valor",
    xaxis_title="Variable"
)

fig.show()
```

```python
# ============================================================
# CELDA 28 - HISTOGRAMA INTERACTIVO DE TOTALPRICE
# ============================================================

# Para que el gráfico sea legible, se limita al percentil 99.
limite_p99 = df["TotalPrice"].quantile(0.99)

df_hist = df[df["TotalPrice"] <= limite_p99].copy()

fig = px.histogram(
    df_hist,
    x="TotalPrice",
    nbins=80,
    title="Distribución de TotalPrice tras limpieza, excluyendo valores por encima del percentil 99"
)

fig.update_layout(
    xaxis_title="TotalPrice",
    yaxis_title="Número de transacciones"
)

fig.show()
```

```python
# ============================================================
# CELDA 29 - PIE CHART: PESO DE UK VS INTERNACIONAL
# ============================================================

df_geo = df.copy()
df_geo["Zona"] = np.where(df_geo["Country"] == "United Kingdom", "United Kingdom", "Internacional")

resumen_geo = df_geo["Zona"].value_counts().reset_index()
resumen_geo.columns = ["Zona", "Transacciones"]

fig = px.pie(
    resumen_geo,
    names="Zona",
    values="Transacciones",
    title="Distribución geográfica del dataset limpio: UK vs internacional",
    hole=0.35
)

fig.update_traces(textinfo="percent+label+value")
fig.show()
```
